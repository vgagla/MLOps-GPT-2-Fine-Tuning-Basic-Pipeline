# Fine-Tunerino: Query the inference API

This notebook sends prompts to the local FastAPI service. First train the model using `main.ipynb`, then start the API in a separate terminal from the repository root:

```bash
pip install -r requirements.txt
uvicorn api:app --host 127.0.0.1 --port 8000 --reload
```

Keep the API bound to localhost for local development; this starter API has no authentication and should not be exposed publicly.

````python
import requests

API_URL = "http://127.0.0.1:8000"

health = requests.get(f"{API_URL}/health", timeout=10)
health.raise_for_status()
health.json()
````

## Send a prompt

Use a completion-style prompt such as `Based on the documents, who is ...?` or begin a sentence the model should continue. GPT-2 is a text-completion model, so it may not follow instructions like a chat assistant.

````python
def ask_model(prompt, max_new_tokens=80, temperature=0.7, top_p=0.95):
    response = requests.post(
        f"{API_URL}/generate",
        json={
            "prompt": prompt,
            "max_new_tokens": max_new_tokens,
            "temperature": temperature,
            "top_p": top_p,
        },
        timeout=180,
    )
    response.raise_for_status()
    return response.json()

result = ask_model("Based on the documents, who is Diogo Delgado Barros?")
print(result["generated_text"])
````

````python
# Change this prompt and run the cell again to try another query.
prompt = "Declaration: "
ask_model(prompt, max_new_tokens=60)
````
